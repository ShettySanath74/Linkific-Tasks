```python

from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

DAY7_DIR = Path(r"C:\Users\Comp\Linkific-Tasks-Upload\Day 7")
DATA_FILE = DAY7_DIR / "data" / "ecommerce_sales.csv"
OUTPUT_DIR = DAY7_DIR / "outputs"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(DATA_FILE)
df["sales"] = df["quantity"] * df["unit_price"]

print("Dataset shape:", df.shape)
print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

display(df.head())
display(df.describe(numeric_only=True))
```

```python

category_sales = (
    df.groupby("category", as_index=False)
    .agg(
        total_sales=("sales", "sum"),
        units_sold=("quantity", "sum"),
        average_sales=("sales", "mean")
    )
    .sort_values("total_sales", ascending=False)
)

display(category_sales)

category_sales.to_csv(
    OUTPUT_DIR / "notebook_ecommerce_category_analysis.csv",
    index=False
)
```
