```python
import numpy as np
print("NumPy version:", np.__version__)
```

```python
a = np.array([10, 20, 30, 40, 50])
print(a)
print(np.zeros(5))
print(np.ones(5))
print(np.arange(0, 10, 2))
print(np.linspace(0, 10, 5))
```

```python
print("First:", a[0])
print("Last:", a[-1])
print("First three:", a[:3])
print("From index 2:", a[2:])
```

```python
matrix = np.arange(1, 10).reshape(3, 3)
print(matrix)
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Elements:", matrix.size)
```

```python
print(a + 5)
print(a * 2)
```

```python
print(a - 5)
print(a / 10)
print(np.square(a))
print(np.sqrt(a))
```

```python
print("Sum:", np.sum(a))
print("Mean:", np.mean(a))
print("Median:", np.median(a))
print("Std:", np.std(a))
print("75th percentile:", np.percentile(a, 75))
```

```python
values = np.array([30, 10, 20, 10, 30, 40])
print("Sorted:", np.sort(values))
print("Unique:", np.unique(values))
print("Unique and counts:", np.unique(values, return_counts=True))
```

```python
print(a[a > 25])
print(a[(a >= 20) & (a <= 40)])
```

```python
print(np.where(a >= 30, "High", "Low"))
print("Positions:", np.where(a >= 30)[0])
print("Fancy indexing:", a[[0, 2, 4]])
```

```python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print("Column totals:", np.sum(matrix, axis=0))
print("Row totals:", np.sum(matrix, axis=1))
```

```python
import time

numbers = np.arange(500_000, dtype=np.float64)

start = time.perf_counter()
result_loop = [x * 2 for x in numbers]
loop_time = time.perf_counter() - start

start = time.perf_counter()
result_numpy = numbers * 2
numpy_time = time.perf_counter() - start

print("Loop time:", loop_time)
print("NumPy time:", numpy_time)
print("Results match:", np.array_equal(result_loop, result_numpy))
print("Speedup:", loop_time / numpy_time if numpy_time > 0 else "N/A")
```
